```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node

    def insert_at_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def display(self):
        current = self.head
        while current:
            print(current.data, end=" ")
            current = current.next
        print()
```

```python
#Insert Element at beginning
def insert_at_beginning(self, data):
    new_node = Node(data)
    new_node.next = self.head
    self.head = new_node

LinkedList.insert_at_beginning = insert_at_beginning
```

```python
def insert_at_end(self, data):
    new_node = Node(data)
    if not self.head:
        self.head = new_node
        return
    
    current = self.head
    while current.next:
        current = current.next
    current.next = new_node
   
LinkedList.insert_at_end = insert_at_end
```

```python
def delete(self, data):
    # Delete from beginning
    if self.head and self.head.data == data:
        self.head = self.head.next
        return
    
    # Delete from middle/end
    current = self.head
    while current and current.next:
        if current.next.data == data:
            current.next = current.next.next
            return
        current = current.next
LinkedList.delete = delete
```

```python
def reverse(self):
    prev = None
    current = self.head
    
    while current:
        next_temp = current.next  # Save next
        current.next = prev        # Reverse
        prev = current             # Move prev
        current = next_temp        # Move current
    
    self.head = prev
LinkedList.reverse = reverse
```

```python
def find_middle(self):
    if not self.head:
        return None
    
    slow = self.head
    fast = self.head
    
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    
    return slow.data
LinkedList.find_middle = find_middle
```

```python
def has_cycle(self):
    if not self.head:
        return False
    
    slow = self.head
    fast = self.head
    
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        
        if slow == fast:
            return True
    
    return False
LinkedList.has_cycle = has_cycle
```

```python
def merge_sorted_lists(head1, head2):
    dummy = Node(0)
    current = dummy
    
    while head1 and head2:
        if head1.data <= head2.data:
            current.next = head1
            head1 = head1.next
        else:
            current.next = head2
            head2 = head2.next
        current = current.next
    
    # Link remaining
    if head1:
        current.next = head1
    else:
        current.next = head2
    
    return dummy.next
LinkedList.merge_sorted_lists = merge_sorted_lists
```

```python
def remove_duplicates(self):
    if not self.head:
        return
    
    current = self.head
    
    while current and current.next:
        if current.data == current.next.data:
            current.next = current.next.next
        else:
            current = current.next
LinkedList.remove_duplicates = remove_duplicates
```

```python
def rotate(self, k):
    if not self.head or k == 0:
        return
    
    # Find length
    length = 0
    current = self.head
    while current:
        length += 1
        current = current.next
    
    k = k % length
    if k == 0:
        return
    
    # Find node at (length - k - 1)
    current = self.head
    for _ in range(length - k - 1):
        current = current.next
    
    # Reconnect
    new_head = current.next
    current.next = None
    
    # Find last node and connect to old head
    last = new_head
    while last.next:
        last = last.next
    last.next = self.head
    
    self.head = new_head
LinkedList.rotate = rotate
```

```python
#Given an array of n-1 integers in the range [1, n], find the missing number.
def find_missing_number(arr):
    n = len(arr) + 1  # Since one number is missing
    total_sum = n * (n + 1) // 2  # Sum of first n natural numbers
    arr_sum = sum(arr)  # Sum of the given array
    missing_number = total_sum - arr_sum  # The missing number is the difference
    return missing_number
arr = [1, 2, 4, 5]
missing_number = find_missing_number(arr)
print("The missing number is", missing_number)
```

```python
#find peak using binary search 
def find_peak(arr):          #peak element is an element which is greater than its neighbors
    left, right = 0, len(arr) - 1
    
    while left < right:
        mid = (left + right) // 2
        
        if arr[mid] < arr[mid + 1]:
            left = mid + 1
        else:
            right = mid
            
    return arr[left]
arr = [1, 3, 20, 4, 1, 0]
peak = find_peak(arr)
print("The peak element is", peak)
```

```python
#rotate an array
def rotate_array(arr, k):      #it rotates the array to the right by k steps, where k is non-negative. For example, if the input array is [1, 2, 3, 4, 5] and k is 2, then the output will be [4, 5, 1, 2, 3].
    k = k % len(arr)  # Handle cases where k is greater than array length
    return arr[-k:] + arr[:-k]
arr = [1, 2, 3, 4, 5]
k = 2
rotated_arr = rotate_array(arr, k)
print("Rotated array:", rotated_arr)
```

```python
#Find all peaks in an array
def find_peaks(arr):
    peaks = []            #peaks are the elements which are greater than their neighbors
    n = len(arr)
    
    for i in range(n):
        if (i == 0 or arr[i] >= arr[i - 1]) and (i == n - 1 or arr[i] >= arr[i + 1]):
            peaks.append(arr[i])
    
    return peaks
arr = [1, 3, 2, 5, 4]
peaks = find_peaks(arr)
print("Peaks in the array:", peaks)
```
