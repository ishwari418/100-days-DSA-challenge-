```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node

    def insert_at_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def display(self):
        current = self.head
        while current:
            print(current.data, end=" ")
            current = current.next
        print()
```

```python
#Insert Element at beginning
def insert_at_beginning(self, data):
    new_node = Node(data)
    new_node.next = self.head
    self.head = new_node

LinkedList.insert_at_beginning = insert_at_beginning
```

```python
def insert_at_end(self, data):
    new_node = Node(data)
    if not self.head:
        self.head = new_node
        return
    
    current = self.head
    while current.next:
        current = current.next
    current.next = new_node
   
LinkedList.insert_at_end = insert_at_end
```

```python
def delete(self, data):
    # Delete from beginning
    if self.head and self.head.data == data:
        self.head = self.head.next
        return
    
    # Delete from middle/end
    current = self.head
    while current and current.next:
        if current.next.data == data:
            current.next = current.next.next
            return
        current = current.next
LinkedList.delete = delete
```

```python
def reverse(self):
    prev = None
    current = self.head
    
    while current:
        next_temp = current.next  # Save next
        current.next = prev        # Reverse
        prev = current             # Move prev
        current = next_temp        # Move current
    
    self.head = prev
LinkedList.reverse = reverse
```

```python
def find_middle(self):
    if not self.head:
        return None
    
    slow = self.head
    fast = self.head
    
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    
    return slow.data
LinkedList.find_middle = find_middle
```

```python
def has_cycle(self):
    if not self.head:
        return False
    
    slow = self.head
    fast = self.head
    
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        
        if slow == fast:
            return True
    
    return False
LinkedList.has_cycle = has_cycle
```

```python
def merge_sorted_lists(head1, head2):
    dummy = Node(0)
    current = dummy
    
    while head1 and head2:
        if head1.data <= head2.data:
            current.next = head1
            head1 = head1.next
        else:
            current.next = head2
            head2 = head2.next
        current = current.next
    
    # Link remaining
    if head1:
        current.next = head1
    else:
        current.next = head2
    
    return dummy.next
LinkedList.merge_sorted_lists = merge_sorted_lists
```

```python
def remove_duplicates(self):
    if not self.head:
        return
    
    current = self.head
    
    while current and current.next:
        if current.data == current.next.data:
            current.next = current.next.next
        else:
            current = current.next
LinkedList.remove_duplicates = remove_duplicates
```

```python
def rotate(self, k):
    if not self.head or k == 0:
        return
    
    # Find length
    length = 0
    current = self.head
    while current:
        length += 1
        current = current.next
    
    k = k % length
    if k == 0:
        return
    
    # Find node at (length - k - 1)
    current = self.head
    for _ in range(length - k - 1):
        current = current.next
    
    # Reconnect
    new_head = current.next
    current.next = None
    
    # Find last node and connect to old head
    last = new_head
    while last.next:
        last = last.next
    last.next = self.head
    
    self.head = new_head
LinkedList.rotate = rotate
```

```python
#Given an array of n-1 integers in the range [1, n], find the missing number.
def find_missing_number(arr):
    n = len(arr) + 1  # Since one number is missing
    total_sum = n * (n + 1) // 2  # Sum of first n natural numbers
    arr_sum = sum(arr)  # Sum of the given array
    missing_number = total_sum - arr_sum  # The missing number is the difference
    return missing_number
arr = [1, 2, 4, 5]
missing_number = find_missing_number(arr)
print("The missing number is", missing_number)
```

```python
#find peak using binary search 
def find_peak(arr):          #peak element is an element which is greater than its neighbors
    left, right = 0, len(arr) - 1
    
    while left < right:
        mid = (left + right) // 2
        
        if arr[mid] < arr[mid + 1]:
            left = mid + 1
        else:
            right = mid
            
    return arr[left]
arr = [1, 3, 20, 4, 1, 0]
peak = find_peak(arr)
print("The peak element is", peak)
```

```python
#rotate an array
def rotate_array(arr, k):      #it rotates the array to the right by k steps, where k is non-negative. For example, if the input array is [1, 2, 3, 4, 5] and k is 2, then the output will be [4, 5, 1, 2, 3].
    k = k % len(arr)  # Handle cases where k is greater than array length
    return arr[-k:] + arr[:-k]
arr = [1, 2, 3, 4, 5]
k = 2
rotated_arr = rotate_array(arr, k)
print("Rotated array:", rotated_arr)
```

```python
#Find all peaks in an array
def find_peaks(arr):
    peaks = []            #peaks are the elements which are greater than their neighbors
    n = len(arr)
    
    for i in range(n):
        if (i == 0 or arr[i] >= arr[i - 1]) and (i == n - 1 or arr[i] >= arr[i + 1]):
            peaks.append(arr[i])
    
    return peaks
arr = [1, 3, 2, 5, 4]
peaks = find_peaks(arr)
print("Peaks in the array:", peaks)
```

```python
#Sliding Window Maximum is a problem which is to find the maximum element in each sliding window of size k in the given array. For example, if the input array is [1, 3, -1, -3, 5, 3, 6, 7] and k is 3, then the output will be [3, 3, 5, 5, 6, 7].
def sliding_window_maximum(nums, k):
    if not nums:
        return []
    
    result = []
    for i in range(len(nums) - k + 1):
        window_max = max(nums[i:i+k])
        result.append(window_max)
    
    return result
nums = [1,3,-1,-3,5,3,6,7]
k = 3
print(sliding_window_maximum(nums, k))  # Output: [3, 3, 5, 5, 6, 7]
```

```python
#Stock Span Problem
nums = [100, 80, 60, 70, 60, 75, 85]
span = [1] * len(nums)  # Initialize span array with 1
stack = []
for i in range(len(nums)):
    while stack and nums[i] >= nums[stack[-1]]:
        stack.pop()
    span[i] = i + 1 if not stack else i - stack[-1]
    stack.append(i)
    print(span)
```

```python
#Next Greater Element is a problem which is to find the next greater element for each element in the array. The next greater element for an element x is the first greater element on the right side of x in the array. If there is no greater element on the right side, then the next greater element for x is -1.
def next_greater_element(nums):
    stack = []
    result = [-1] * len(nums)
    for i in range(len(nums)):
        while stack and nums[i] > nums[stack[-1]]:
            result[stack.pop()] = nums[i]
        stack.append(i)
    return result
nums = [4, 5, 2, 10]
print(next_greater_element(nums))
```

```python
#Find Duplicate Element
def find_duplicate(arr):
    seen = set()
    for num in arr:
        if num in seen:
            return num
        seen.add(num)
    return None
arr = [1, 2, 3, 4, 2]
print(find_duplicate(arr))
```

```python
#Daily Temperatures
def daily_temperatures(T):
    stack = []
    result = [0] * len(T)
    for i in range(len(T)):
        while stack and T[i] > T[stack[-1]]:
            index = stack.pop()
            result[index] = i - index
        stack.append(i)
    return result
T = [73, 74, 75, 71, 69, 72, 76, 73]
print(daily_temperatures(T))
```

```python
#two sum problem
def two_sum(nums, target):
    num_map = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in num_map:
            return [num_map[complement], i]
        num_map[num] = i
    return None
nums = [2, 7, 11, 15]
target = 9
print(two_sum(nums, target))  # Output: [0, 1] (since nums[0] + nums[1] = 2 + 7 = 9)
```

```python
#find piot element in the array
def find_pivot(arr):
    left_sum = 0
    total_sum = sum(arr)
    
    for i, num in enumerate(arr):
        total_sum -= num  # Now total_sum is the right sum
        if left_sum == total_sum:
            return i  # Pivot index found
        left_sum += num
    
    return -1  # No pivot index found
```

```python
#3 sum problem is a problem which is to find all unique triplets in the array which gives the sum of zero. For example, if the input array is [-1, 0, 1, 2, -1, -4], then the output will be [[-1, -1, 2], [-1, 0, 1]].
def three_sum(nums):
    nums.sort()
    result = []
    
    for i in range(len(nums) - 2):
        if i > 0 and nums[i] == nums[i - 1]:
            continue  # Skip duplicates
        
        left, right = i + 1, len(nums) - 1
        while left < right:
            total = nums[i] + nums[left] + nums[right]
            if total < 0:
                left += 1
            elif total > 0:
                right -= 1
            else:
                result.append([nums[i], nums[left], nums[right]])
                while left < right and nums[left] == nums[left + 1]:
                    left += 1  # Skip duplicates
                while left < right and nums[right] == nums[right - 1]:
                    right -= 1  # Skip duplicates
                left += 1
                right -= 1
                
    return result
a=three_sum([-1, 0, 1, 2, -1, -4])
print(a)  # Output: [[-1, -1, 2], [-1
```

```python
#Merge Intervals
def merge_intervals(intervals):
    if not intervals:
        return []
    
    intervals.sort(key=lambda x: x[0])  # Sort by start time
    merged = [intervals[0]]
    
    for current in intervals[1:]:
        last_merged = merged[-1]
        if current[0] <= last_merged[1]:  # Overlapping intervals
            last_merged[1] = max(last_merged[1], current[1])  # Merge
        else:
            merged.append(current)  # No overlap, add to result
            
    return merged
a=merge_intervals([[1,3],[2,6],[8,10],[15,18]])
print(a)  # Output: [[1, 6], [8, 10], [15, 18]]
```

```python
#Find Peak Element
def find_peak_element(nums):
    left, right = 0, len(nums) - 1
    
    while left < right:
        mid = (left + right) // 2
        if nums[mid] < nums[mid + 1]:
            left = mid + 1
        else:
            right = mid
            
    return left
a=find_peak_element([1, 2, 3, 1])
print(a)  # Output: 2 (The peak element is 3 at index 2)
```

```python
#Valid Parentheses
def is_valid_parentheses(s):
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}
    
    for char in s:
        if char in mapping:
            top_element = stack.pop() if stack else '#'
            if mapping[char] != top_element:
                return False
        else:
            stack.append(char)
    
    return not stack
a=is_valid_parentheses("()[]{}")
print(a)  # Output: True (The input string is valid)
```

```python
#segment tree is a data structure that allows answering range queries and updating elements in logarithmic time. It is particularly useful for problems involving intervals or segments of an array. A segment tree is built on top of an array, where each node represents a segment (or range) of the array, and the value of each node is typically a function (like sum, min, max) of the elements in that segment.
#segment tree problem
def build_segment_tree(arr):
    n = len(arr)
    tree = [0] * (2 * n)
    
    # Build the tree by inserting leaves
    for i in range(n):
        tree[n + i] = arr[i]
    
    # Build the rest of the tree by calculating parents
    for i in range(n - 1, 0, -1):
        tree[i] = tree[2 * i] + tree[2 * i + 1]
    
    return tree
```

```python
#two pointer technique is a common algorithmic approach used to solve problems involving arrays or linked lists. It involves using two pointers (or indices) to traverse the data structure, often with one pointer moving faster than the other. This technique is particularly useful for problems that require finding pairs, subarrays, or specific conditions within a sorted array or linked list.
def two_pointer_technique(arr, target):
    left, right = 0, len(arr) - 1
    
    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            return (left, right)  # Found the pair
        elif current_sum < target:
            left += 1  # Move left pointer to the right
        else:
            right -= 1  # Move right pointer to the left
            
    return None  # No pair found
```

```python
#sliding window technique is a common algorithmic approach used to solve problems involving arrays or strings. It involves maintaining a "window" of elements that satisfies certain conditions and moving this window through the data structure to find optimal solutions. This technique is particularly useful for problems that require finding subarrays, substrings, or specific conditions within a contiguous segment of the data structure.
def sliding_window_technique(arr, k):
    if not arr or k <= 0:
        return []
    
    window_sum = sum(arr[:k])
    max_sum = window_sum
    result = [max_sum]
    
    for i in range(k, len(arr)):
        window_sum += arr[i] - arr[i - k]  # Slide the window
        max_sum = max(max_sum, window_sum)
        result.append(max_sum)
    
    return result
```

```python
#Prefix sum 
arr = [2,3,5,1]

prefix = [0]*len(arr)

prefix[0] = arr[0]

for i in range(1,len(arr)):
    prefix[i] = prefix[i-1] + arr[i]

print(prefix)
```

```python
arr = [2,3,5,1,4]

prefix = [0]*len(arr)

prefix[0] = arr[0]

for i in range(1,len(arr)):
    prefix[i] = prefix[i-1] + arr[i]

left = 1
right = 3

if left == 0:
    print(prefix[right])
else:
    print(prefix[right]-prefix[left-1])
```

```python
#subarray
arr=[2,-1,3,-2,4]
current_sum=0
max_sum=0
for num in arr:
    current_sum=current_sum+num
    if current_sum>max_sum:
        max_sum=current_sum
        if current_sum<0:
            current_sum=0
print(max_sum)
```

```python
arr=[3,5,6,8,7,9]
frequency={}
for num in arr:
    if num in frequency:
        frequency[num]+=1
    else:
        frequency[num]=1
print(frequency)
```

```python
arr=[3,4,5,3,7,5]
seen={}
for num in arr:
    if num in seen:
        print(True)
        break
    seen[num]=True
else:
    print(False)
```

```python
arr = [2, 7, 11, 15,5]
target = 12

seen = {}

for i in range(len(arr)):

    need = target - arr[i]

    if need in seen:
        print([seen[need], i])
        break

    seen[arr[i]] = i
```

```python
word1 = "listen"
word2 = "silentS"

count1 = {}
count2 = {}

for ch in word1:
    count1[ch] = count1.get(ch, 0) + 1

for ch in word2:
    count2[ch] = count2.get(ch, 0) + 1

if count1 == count2:
    print(True)
else:
    print(False)
```
