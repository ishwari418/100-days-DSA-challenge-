```python
num=(int(input("Enter the number: ")))
l=len(str(abs(num)))   #abs does not take negative sign 
sum=0
i=num
while i>0:
    digit=i%10
    sum+=digit**3
    i//=10
if sum==num:
    print (num, "is an armstrong number")
else:
    print (num, "is not armstrong number")
```

```python
number=int(input("Enter the number"))
sum_of_factorial=0
for i in range (1, number):
    if number%i==0:
        sum_of_factorial+=i

if sum_of_factorial==number:
    print (number, "is the perfect number")
else:
    print (number, "is not perfect number")
```

```python
a=int (input("Enter the number"))
b=int (input("Enter the 2nd number"))
greater=max(a,b)
while True:
    if greater%a==0 and greater % b==0:
        lcm=greater
        break
    greater+=1
print ("The LCM of ", a, "and ", b, "is", lcm)
```

```python
num=int(input("Enter the number"))
reverse=0
while num >0:
    digit =num%10
    reverse=reverse*10+digit
    num//=10
print ("reverse number is :", reverse)
```

```python
#palidrome number
num=int(input("Enter the number"))
reverse=0
temp=num
while num > 0:
    digit =num%10
    reverse=reverse*10+digit
    num//=10
if temp==reverse:
    print (temp, "is a palindrome number")
else:
    print (temp, "is not a palindrome number")
```

```python
#find largest number in the list
numbers=[1,2,3,4,5]
largest=numbers[0]
for i in numbers:
    if i>largest:
        largest=i
print ("The largest number in the list is", largest)
```

```python
#find smallest number in the list
numbers=[1,2,3,4,5]
smallest=numbers[0]
for i in numbers:
    if i<smallest:
        smallest=i
print ("The smallest number in the list is", smallest)
```
