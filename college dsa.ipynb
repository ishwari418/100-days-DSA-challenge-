```python
from traitlets import List


class solution:
    def fizzBuzz(self, n: int) -> List[str]:
        ans = []
        for i in range(1, n + 1):
            if i % 3 == 0 and i % 5 == 0:
                ans.append("FizzBuzz")
            elif i % 3 == 0:
                ans.append("Fizz")
            elif i % 5 == 0:
                ans.append("Buzz")
            else:
                ans.append(str(i))
        return ans
s = solution()
print(s.fizzBuzz(15))
```

```python
def countodds(self. low:int, high:int) -> int:
    if low % 2 == 0:
        low += 1
    if high % 2 == 0:
        high -= 1
    if low > high:
        return 0
    return (high - low) // 2 + 1    .
```
